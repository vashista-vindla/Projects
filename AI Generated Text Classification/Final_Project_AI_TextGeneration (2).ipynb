```python
# Binning the 'Value' column

bins = range(0,60,5)
labels = []  # Define bin edges
for i in range(0,60,5):
  labels.append(str(i))
labels.pop(0)

df['Value_Bin'] = pd.cut(df['word_count'], bins=bins, labels=labels, include_lowest=True)

# Group by 'Category' and the new 'Value_Bin' column, then perform aggregation
result = df.groupby(['ind', 'Value_Bin']).size().reset_index(name='Count')

# Display the result
display(result)
```

* In the above table,the number of words in the scentences have been grouped into bins of 5.
* It was observed that for human generated text, maximum number of scentences fall in the category of 30-35.
* In a similar fashion, maximum number of scentences for AI generated text fall in the catergory of 25-30.

## Data Preparation for Modeling


```python
df.columns
```

Let's check if there are any missing values in the dataframe

```python
# pd.set_option('display.max_columns', None)
df.isna().sum()
```

So there is no missing values in the dataframe so we don't need to impute the values

We have to split the model into train and test dataframes, where the test dataframe will only have the target variable `ind` and the train dataframe will have all the columns except the target variable and `ID` columns

We are using X to get the train dataframe and y for the test dataframe

```python
from sklearn.model_selection import train_test_split
X=df.drop(['ID','ind'],axis=1) # Dropping the target variable and ID column
y=df['ind'] # Storing the target column in a variable
```

```python
print(X.shape,y.shape) # Checking the shape of the train and test dataframe
```

By using `train_test_split()` function we are splitting `X` and `y` to `X_train`,`X_test`,`y_train`,and `y_test` with `test_size=0.1`

```python
#Splitting the data

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.1,random_state=42) #(setting random state = 42 and using 90/10 split)
print(X_train.shape,X_test.shape,y_train.shape,y_test.shape)# Checking the shape of the four dataframes
```

## Model Building
Once we are done with splitting the dataframes, let us build our model

```python
#Install and import the required modules
!pip install scikeras
from scikeras.wrappers import KerasClassifier
from tensorflow.keras.utils import to_categorical

# for modeling
from keras.models import Sequential
from keras.layers import Dense, Dropout
from keras.callbacks import EarlyStopping
from tensorflow.keras.layers import Conv1D
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv1D, Flatten
```

```python
# A sequential Model
# Using simple dense layers starting with 769 inputs for the first layer and sigmoid activation fuction in the model

model = Sequential()
model.add(Dense(769, input_shape=(X_train.shape[1],), activation='relu'))
model.add(Dense(700,activation='relu'))
model.add(Dense(650,activation='relu'))
model.add(Dense(600,activation='relu'))
model.add(Dense(550,activation='relu'))
model.add(Dense(500,activation='relu'))
model.add(Dense(450,activation='relu'))
model.add(Dense(400,activation='relu'))
model.add(Dense(350,activation='relu'))
model.add(Dense(300,activation='relu'))
model.add(Dense(250,activation='relu'))
model.add(Dense(200,activation='relu'))
model.add(Dense(150,activation='relu'))
model.add(Dense(105,activation='relu'))
model.add(Dense(75,activation='relu'))
model.add(Dense(50,activation='relu'))
model.add(Dense(25,activation='relu'))
model.add(Dense(20,activation='relu'))
model.add(Dense(15,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(5,activation='relu'))
model.add(Dense(1, activation='sigmoid')) # output
model.summary() # Getting the summary of the model
```

```python
# compile the model

model.compile(optimizer='adam', #Can use rmsprop as well but adam is giving the best results(f1 score)
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

```python
# Using early stopiing for preventing under or over fitting

es = EarlyStopping(monitor='accuracy',
                                   mode='max',
                                   patience=70,
                                   restore_best_weights=True)

# Fitting the Model
history = model.fit(X_train,
                    y_train,
                    callbacks=[es],
                    epochs=1000000,
                    batch_size=100,
                    validation_split=0.1,
                    shuffle=True,
                    verbose=1)
```

## Model Evaluation

Once the model had learned the patterns from the train dataset, we can check on how it learned itself by plotting the accuracy on train data

```python
# learning curve
import matplotlib.pyplot as plt

# accuracy
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']

# loss
loss = history.history['loss']
val_loss = history.history['val_loss']

epochs = range(1, len(acc) + 1)

# "bo" is for "blue dot"
plt.plot(epochs, acc, 'bo', label='Training accuracy')
# b is for "solid blue line"
plt.plot(epochs, val_acc, 'b', label='Validation accuracy')
plt.title('Training and validation accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```

The graph shows us that for training data from the training dataframe it learned perfectly and for validation data it fluctuates from 0.89 to 0.94

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
import numpy as np

preds = np.round(model.predict(X_test),0)

# confusion matrix
confusion_matrix(y_test, preds)

```

```python
print(classification_report(y_test, preds))
```

We can see that our total accuracy is 0.94 and it is a good model

```python
# Obtaining F1 score for checking Model performance

from sklearn.metrics import f1_score
f1_score(y_test,preds)
```
